# ToolCallingAgent

`ToolCallingAgent` is an agent who rely on the LLM natvie ability to call functions/API. The LLM is give each tool's JSON schema, eg. name, description, parameter type, result type, etc. The LLM uses this information to call the functions.

```python
# TODO: Create a tool calling agent
agent_model = OpenAIServerModel(model_id=model, api_key=OPENAI_API_KEY, reasoning_effort='none')
tool_calling_agent = ToolCallingAgent(model=agent_model, max_steps=8
                , tools=[get_latlng, get_temperature, query_album_track], add_base_tools=False)


```

# Memory Retention
`reset` parameter is use to control the agent's memory retention between consecutive runs. Every time an agent is executed `run()`, it will forget all past conversation. It handles the `run()` as if it is the first request. This this the default behaviour `reset=True`.

By setting `reset=False`, we can ask the agent to use past conversation as context for the current conversation. This is essentially a follow-up quesiton paradigm.

```python
# TODO: Ask 2 related questions. Set reset=True (default) when asking the second question. Perform the evaluation again by setting reset=False
tool_calling_agent.run("List all the prime numbers from 1 to 100 inclusive")
```

```python
tool_calling_agent.run("What is the sum of the prime numbers?", reset=False)
```
