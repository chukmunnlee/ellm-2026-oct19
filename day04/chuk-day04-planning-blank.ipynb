# Planning

Agent runs in a ReAct loop: thought -> action -> observation.

The `planning_interval=n` contols how frequently the agent pauses at every `n` steps to plan by making an extra call to LLM. This pause is use to relect on the task:
- Initial planning when the agent start - the model write out a step-by-step plan for the task
- At every n step - the model look at what has happened so far and update/revises its plan.
The plan is stored in the agents `PlanningStep`.

```python
# TODO: Create ToolCallingAgent with `planning_interval`
tool_calling_agent = ToolCallingAgent(model=agent_model, planning_interval=3, add_base_tools=True, tools=[])
```

```python
# List the planning steps
def print_agent_steps(agent: ToolCallingAgent):
   for step in agent.memory.steps:
      if isinstance(step, PlanningStep):
         print('**** PLAN')
         print(f'{step.plan}')
         print(f'token: {step.token_usage}')
      elif isinstance(step, FinalAnswerStep):
         print('**** OUTCOME')
         print(f'{step.output}')
```

# Human In The Loop

We can review an agent's plan (approve, edit, reject) before the agent act on it. Associate a review function with `PlanningStep` and the agent will use the function as callback to solicit user's approval.

`agent.write_memory_to_messages()` save the agent's memory/step as LLM messages viz. creating context for the agent.

```python
# Review function
def review_plan(step: PlanningStep, agent: ToolCallingAgent) -> None:
   print(f"**** Proposed plan: {step.plan}")
   stop = False
   while not stop:
      choice = input("[Aa]pprove, [Ee]dit, [Rr]eject: ").strip().lower()
      stop = choice in {"a", "e", "c"}
      if "e" == choice:
         # NOTE: Since we are editing, we want to save what was said before
         agent.write_memory_to_messages()
         print("Write your new plan. Enter a new empty line to end")
         lines = iter(input, "")
         step.plan = "**** PLAN UPDATE\n\n" + "\n".join(lines) + "\n**** END OF PLAN UPDATE\n" + step.plan
         print(f"**** USER UPDATE ****\n{step.plan}\n************************")
      elif "r" == choice:
         raise Exception("User rejected this plan")
```

```python
#TODO: Create an agent with review
tool_calling_agent = ToolCallingAgent(model=agent_model
        , planning_interval=3, add_base_tools=True, tools=[]

        , step_callbacks={ PlanningStep: review_plan })

```
