## T5 Models

The <code>flan-t5</code> is a Text-To-Text Transfer Transformer (T5) that is capable of performing zero-shot NLP task such as summary, simple reasoninig, answering questions, etc.

Some T5 models from Huggingface
- [<code>google/flan-t5-base</code>](https://huggingface.co/google/flan-t5-base)

```python
model_name = 'google/flan-t5-base'
#model_name = 'sshleifer/distilbart-cnn-12-6'
```

```python
# TODO: Load tokenizer and model
# Load the tokenizer
tokenizer = AutoTokenizer.from_pretrained(model_name)
# Load the model
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)
```

```python
# TODO: Print the model
print(model)
```

## Configuration
- Set the temperate

```python
# TODO: Configure the LLM
config = GenerationConfig(
    do_sample=True, # temperature, top_k, top_p
    temperature=3.0,
    max_new_tokens=200,
    num_beams=3
)
```

```python
article = "Indonesia's wildfires are not only burning swaths of jungle, plantation and farmland. They are also threatening the country's four recently launched projects that issue carbon credits worth hundreds of millions of dollars by preventing greenhouse gas emissions from 225,000 hectares of forest."
prompt = f"Article:\n\n{article}\nWhat is a summary?"
print(prompt)
enc_prompt = tokenizer(prompt, return_tensors='pt')

# Pass the config to the model to configure the generation
enc_summary = model.generate(**enc_prompt, generation_config=config)

summary = tokenizer.decode(enc_summary[0], skip_special_tokens=True)
print(summary)

```
